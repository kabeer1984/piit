```python
##MAP Function
```

```python
strings = ["my","world","apple","pear"]
lengths = map(len,strings)
```

```python
print (list(lengths))
```

```python
len(list("my"))
```

```python
lengths = map(lambda x: x + "s",strings)
print(list(lengths))
```

```python
lengths = map(lambda x: x + "x",strings)
print(list(lengths))
```

```python
#FILTER Function
```

```python
#custom function
```

```python
def longer_than_4(string):
    return len(string)>4
```

```python
strings = ["my","world","apple","pear"]
```

```python
filtered = filter(longer_than_4,strings)
print(list(filtered))
```

```python
#also by lambda
```

```python
strings = ["my","world","apple","pear"]
```

```python
filtered = filter(lambda x : len(x)>4,strings)
print(list(filtered))
```

```python
print(list(strings))
```

```python
##SORTED Function
```

```python
numbers = [2,4,3,8,7,45,33,82,345,78779]
sorted_nums = sorted(numbers)
print(sorted_nums)
```

```python
numbers = [2,4,3,8,7,45,33,82,345,78779]
sorted_nums = sorted(numbers, reverse=True)
print(sorted_nums)
```

```python
##ENUMERATE Function
```

```python
tasks = ["Write report","Attend meeting","Review code","Submit timesheet"]
```

```python
for index in range(len(tasks)):
    task = tasks[index]
```

```python
print(f"{index+1},{tasks}")
```

```python
for index,task in enumerate(tasks):
    print (f"{index + 1},{task}")
```
