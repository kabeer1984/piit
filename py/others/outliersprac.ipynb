```python
##IQR (DIFFERENCE OF Q3 & Q1)
```

```python
x7=[10,20,30,40,50,60,70]
Q1 = np.percentile(x7,25)
Q3 = np.percentile(x7,75)
IQR = Q3-Q1
print(x7,"\n Q1:", Q1,"Q3", Q3,"IQR", IQR)
```

```python
from scipy.stats import iqr
import  numpy as np
```

```python
x7B = np.array([10,20,30,40,50,60,70])
IQR = iqr(x7B)
print(x7B,"\n IQR", IQR)
```

```python
import statistics as stats

x7c = [10, 20, 30, 40, 50, 60, 70]

# Quartiles: Q1 = q[0], Q2 = q[1], Q3 = q[2]
q = stats.quantiles(x7c, n=4)

# IQR = Q3 - Q1
IQR = q[2] - q[0]

print(x7c, "\nQ3:", q[2], "Q1:", q[0], ":: IQR:", IQR)
```
