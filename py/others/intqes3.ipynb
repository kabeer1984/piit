16.Write a query to calculate the percentage of total sales for each category.

sql:SELECT category,
       SUM(sales) AS category_sales,
       SUM(sales) * 100.0 / (SELECT SUM(sales) FROM sales_table) AS sales_percentage
FROM sales_table
GROUP BY category;

python:category_sales = df.groupby('category')['sales'].sum().reset_index()
total_sales = df['sales'].sum()
category_sales['sales_percentage'] = (category_sales['sales'] / total_sales) * 100

r:library(dplyr)
category_sales <- df %>%
  group_by(category) %>%
  summarise(category_sales = sum(sales)) %>%
  mutate(sales_percentage = (category_sales / sum(category_sales)) * 100)
