```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```


# Load real mtcars dataset from seaborn

```python
df = sns.load_dataset("mpg").dropna()
df.head()
```

###This dataset contains:

mpg — miles per gallon (fuel efficiency)

cylinders — number of engine cylinders

##check & drop missing values

```python
df = df.dropna()
```

##. Exploratory Visualizations

##“Are most cars fuel efficient, or is fuel economy spread out?”

```python
plt.figure(figsize=(8,5))
sns.histplot(df['mpg'], kde=True)
plt.title("Distribution of Fuel Efficiency (MPG)")
plt.show()
```

##Story Insight:
Fuel efficiency varies widely, but many cars cluster around 20–30 MPG.

##Relationship Between Weight and MPG

“Do heavier cars burn more fuel?”

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='weight', y='mpg')
plt.title("Weight vs MPG")
plt.show()
```

Story Insight:
A clear downward trend appears — heavier cars get worse gas mileage.

Cylinders and Fuel Efficiency

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='cylinders', y='mpg')
plt.title("Fuel Efficiency Across Cylinder Types")
plt.show()
```

Story Insight:
More cylinders = lower fuel efficiency.
4-cylinder cars are significantly more efficient than 6 or 8-cylinder cars.
