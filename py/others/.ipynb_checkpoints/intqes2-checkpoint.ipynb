4.	How do you handle missing values in pandas?                                                                I handle missing values in pandas by first identifying them using isnull() or isna(). Then, based on the data type and business context, I decide whether to remove or impute them. For numeric columns, I typically use the mean or median; for categorical columns, I use the mode or a placeholder like ‘Unknown’; and for time-series data, I apply forward or backward fill. After cleaning, I always validate the results to ensure data quality is maintained.
5.	What’s the difference between .loc[] and .iloc[]?                                                        .loc[] is label-based indexing — you select rows and columns by their names/labels.
.iloc[] is position-based indexing — you select rows and columns by their integer positions.             df.loc[0, 'sales']    # by label
df.iloc[0, 2]         # by position
6.	How do you merge two DataFrames in pandas?                                                                  I merge two DataFrames in pandas using the merge() function, typically joining them on a common key. I can control the type of join—such as inner, left, right, or outer—depending on the business requirement. This allows me to combine data accurately while preserving or filtering records as needed.                           ex: pd.merge(df1, df2, on='id', how='inner')
7.	Explain the use of groupby() in pandas.                                                               groupby() is used to group data by one or more columns and apply aggregations like sum(), mean(), or count() to analyze patterns within groups.

10.	What’s the difference between apply() and map()?                                                            map() → element-wise operation on a Series; simple transformations.
apply() → works on Series or DataFrame; can do row-wise, column-wise, or complex operations.
Simple rule: map() = single-column, apply() = flexible/multi-column

12.	How do you filter rows based on multiple conditions in pandas?                                              Use boolean conditions with & (and) or | (or) inside .loc[].
Example:

Always wrap each condition in parentheses.

It’s preferred because it’s faster, more efficient, and leverages optimized C/NumPy code, unlike slow Python loops.
