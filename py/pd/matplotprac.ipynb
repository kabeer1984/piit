```python
from pydataset import data
import pandas as pd
import matplotlib.pyplot as plt
```

```python
mtcars= data('mtcars')
mtcars.head()
```

```python
plt.hist(mtcars['mpg'])
plt.xlabel('speed')
plt.ylabel('wt')
plt.title('Histogram of Mtcars')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

bins = np.histogram_bin_edges(mtcars['mpg'], bins=10)
hist, edges = np.histogram(mtcars['mpg'], bins=bins)

plt.bar(edges[:-1], hist, width=np.diff(edges))

plt.xlabel('mpg')
plt.ylabel('Frequency')
plt.title('Bar Chart of mpg (Binned)')
plt.show()
```

```python
from pydataset import data
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
mtcars= data ('mtcars')
mtcars.head()
```

```python
sns.countplot(x='mpg', data=mtcars)

plt.xlabel('MPG')
plt.ylabel('Count')
plt.title('Bar Chart of MPG Counts (Seaborn)')
plt.show()
```

```python
sns.countplot(x='hp', data=mtcars)

plt.xlabel('hp')
plt.ylabel('Count')
plt.title('Bar Chart of MPG Counts (Seaborn)')
plt.show()
```
